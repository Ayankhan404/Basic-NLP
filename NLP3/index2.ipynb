### Bag of Words

```python
import pandas as pd 
import numpy as np
```

```python
df = pd.DataFrame({'text':['people watch campusx','campusx watch campusx','people write comments','campusx  write comments'],'output':[1,1,0,0]})
```

```python
df
```

```python
from sklearn.feature_extraction.text import CountVectorizer
```

```python
cv = CountVectorizer()
```

```python
Bow= cv.fit_transform(df['text'])
```

```python
Bow.toarray()
```

```python
## index bata raha hai ki words akaha aa rahe hai
cv.vocabulary_
```

```python
# Get the vocabulary (unique words)
print(cv.get_feature_names_out())
```

### N-Grams

```python
Bi_gram=CountVectorizer(ngram_range=(2,2))
```

```python
ngram=Bi_gram.fit_transform(df['text'])
```

```python
Bi_gram.vocabulary_
```

```python
ngram.toarray()
```
