```python
# remove url example 
def remove_url(text):
    pattern=re.compile(r'https?://\S+|www\.\S+')
    return pattern.sub(r'',text)
```

## Remove punctuation

```python
import string
import time

print(string.punctuation)
```

```python
exclude=string.punctuation
```

```python
def remove_puch(text):
    for char in exclude:
        text=text.replace(char,'')
    return text    
```

```python
text='string. with. punchuation?'
```

```python
remove_puch(text)
```

```python
# second method is the fastest method 

def remove_puch1(text):
    return text.translate(str.maketrans('','',exclude))
```

```python
remove_puch1(text)
```

```python
from nltk.corpus import stopwords
```

```python
stopwords.words('English')
```

```python
def remove_stopwords(text):
    new_text=[]
    
    for word in text.split():
        if word in stopwords.words('english'):
            new_text.append('')
        else:
            new_text.append(word)
            
    x=new_text[:]
    new_text.clear()
    return " ".join(x)            
```

```python
remove_stopwords('my name is ayan khan and i am form dholpur rajasthan and i am pursuing my b.tech from global institute of technology')
```
