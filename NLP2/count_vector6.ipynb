```python
import pandas as pd 
```

```python
df = pd.DataFrame({"name":x})
df
```

```python
from sklearn.feature_extraction.text import CountVectorizer
```

```python
cv = CountVectorizer()
```

```python
new_data =cv.fit_transform(["name"]).toarray()
```

```python
new_data
```

```python
cv.vocabulary_
```

```python
# eg of vectorizer

#boject
v = CountVectorizer()

# sample data 
data=[" love Data Science", " love Machine Learning"]

# fir and transform
vectorzied_data = v.fit_transform(data)

# Get the vocabulary (unique words)
print(v.get_feature_names_out())

# convert the result to array
print(vectorzied_data.toarray())
```
