```python
from tensorflow.keras.preprocessing.text import Tokenizer
```

```python
sentences = [
    'i love my dog',
    'i love my cat',
    'i love my dog!'
]
```

```python
token =Tokenizer(num_words=100,oov_token='<OOV>')
```

```python
token.fit_on_texts(sentences)
```

```python
index =token.word_index
```

```python
sequences =token.texts_to_sequences(sentences)
```

```python
print(sequences)
```
